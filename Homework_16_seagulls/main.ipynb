```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import torch

device = torch.accelerator.current_accelerator() if torch.accelerator.is_available() else torch.device("cpu")
epochs = 150
val_frac = 0.2

device
```

```python
from split_data import split_data
from pathlib import Path

split_data(Path("data"), val_frac=val_frac)
```

```python
from ultralytics import YOLO

model = YOLO("yolo11m.pt")
model.to(device)
```

```python
model.train(
    data="dataset.yaml",
    epochs=epochs,
    patience=50,
    imgsz=640,
    batch=16,
    flipud=0.5,
    scale=0.9,
    copy_paste=0.3,
    mixup=0.1,
    close_mosaic=20,
)
```
