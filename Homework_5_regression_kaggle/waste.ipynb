# [Games Rating](https://www.kaggle.com/competitions/games-rating/overview)

```python
import numpy as np
import pandas as pd
```

```python
import opendatasets as od

dataset_url = 'https://www.kaggle.com/competitions/games-rating'
od.download(dataset_url)
```

```python
train = pd.read_csv('./games-rating/train_data.csv', index_col=0)
test = pd.read_csv('./games-rating/test_data.csv', index_col=0)
```

```python
train = train.drop(columns=["Name", "Mechanics", "Domains"])
test = test.drop(columns=["Name", "Mechanics", "Domains"])

train["Complexity Average"] = train["Complexity Average"].str.replace(',', '.').astype(float)
test["Complexity Average"] = test["Complexity Average"].str.replace(',', '.').astype(float)
train["Rating Average"] = train["Rating Average"].str.replace(',', '.').astype(float)
```

```python
from sklearn.impute import SimpleImputer

target = train["Rating Average"]
features = train.drop(columns=["Rating Average"])

imputer = SimpleImputer(strategy='mean')
train = pd.DataFrame(imputer.fit_transform(features), columns=features.columns)

train["Rating Average"] = target.values

test = pd.DataFrame(imputer.transform(test), columns=test.columns)
```

```python
from sklearn.linear_model import LinearRegression

linreg = LinearRegression()
linreg.fit(train.drop(columns=["Rating Average"]), train["Rating Average"])
result = linreg.predict(test)
```

```python
pd.Series(result, name="Rating Average").reset_index().to_csv('./games-rating/result.csv', index=False)
```
