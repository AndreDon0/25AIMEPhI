```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import json
import random

N = 4050  # Кол-во записей, аналогично вашему файлу
data = {str(i): random.choice([True, False]) for i in range(N)}
with open('./data/submit_random.json', 'w', encoding='utf-8') as f:
    json.dump(data, f, ensure_ascii=False, indent=2)
```

```python
import json

with open('/home/andredon/projects/University/25AIMEPhI/Im_professional/12/data/abstracts.json', 'r', encoding='utf-8') as f:
    data = json.load(f)

result = {k: ('**' in v or '\\(' in v) for k, v in data.items()}
with open('./data/submit.json', 'w', encoding='utf-8') as f:
    json.dump(result, f, ensure_ascii=False, indent=2)
```

```python
true_count = sum(result.values())
false_count = len(result) - true_count

# Данные для графика
labels = ['True', 'False']
counts = [true_count, false_count]

# Визуализация c помощью seaborn
sns.barplot(x=labels, y=counts)
plt.title('True/False counts')
plt.ylabel('Count')
plt.show()
```
