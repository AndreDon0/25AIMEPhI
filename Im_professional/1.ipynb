```python
import numpy as np
from scipy.optimize import minimize
```

```python
x0 = [0.25, 0.25, 0.25, 0.25]
bnds = ((0, 1), (0, 1), (0, 1), (0, 1))
cons = ({'type': 'eq'  , 'fun': lambda x:  x[0] - x[1]},
        {'type': 'eq', 'fun': lambda x:  x[2] - 2*x[3]},
        {'type': 'eq', 'fun': lambda x:  x[0] + x[1] + x[2] + x[3] - 1})
fun = lambda x: sum(x*np.log2(x))

res = minimize(fun, x0, method="trust-constr", bounds=bnds, constraints=cons)
print(res.x)
```
